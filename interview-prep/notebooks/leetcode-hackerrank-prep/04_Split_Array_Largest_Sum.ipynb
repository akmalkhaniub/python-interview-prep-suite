# Split Array Largest Sum (Task Distribution)

**Problem Statement:**
Given an integer array `tasks` (representing the time each task takes) and an integer `k` (representing team members), split `tasks` into `k` non-empty continuous subarrays such that the largest sum among these `k` subarrays is minimized.

Imagine you have a list of tasks that must be done in order, and you want to distribute them among `k` workers such that the worker with the heaviest load has the minimum possible load.


## Approach 1: Dynamic Programming (Suboptimal for large inputs)
**Time Complexity:** O(k * N^2)
**Space Complexity:** O(N * k)
Calculate the prefix sums and use a 2D DP table where `dp[i][j]` is the minimum largest subarray sum for splitting the first `j` tasks into `i` parts.


## Approach 2: Binary Search on Answer (Optimal)
**Time Complexity:** O(N * log(Sum - Max))
**Space Complexity:** O(1)
The minimum possible load is the max element in `tasks` (if we have `N` workers). The maximum possible load is the sum of all `tasks` (if we have 1 worker).
We can binary search the answer `mid` between `[max(tasks), sum(tasks)]`. For each `mid`, we check if it's possible to split the tasks into `k` parts such that no part exceeds `mid`.


```
def splitArray(tasks, k):
    def can_split(max_load):
        current_load = 0
        workers_needed = 1
        for task in tasks:
            if current_load + task > max_load:
                workers_needed += 1
                current_load = task
            else:
                current_load += task
        return workers_needed <= k

    left, right = max(tasks), sum(tasks)
    res = right
    while left <= right:
        mid = left + (right - left) // 2
        if can_split(mid):
            res = mid
            right = mid - 1
        else:
            left = mid + 1
    return res

# Test
print('Minimum max load:', splitArray([7,2,5,10,8], 2)) # Expected 18
```
