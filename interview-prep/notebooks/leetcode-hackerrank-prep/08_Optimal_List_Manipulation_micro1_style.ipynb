## 1️⃣ Pattern: Two Pointers (In-Place Modification)
**Problem:** Given an integer array `nums`, move all `0`s to the end of it while maintaining the relative order of the non-zero elements.
**Constraint:** You must do this in-place without making a copy of the array.

**Approach:** Use a `left` pointer to track the placement of non-zero elements, and a `right` pointer to scan.

````python
def move_zeroes(nums):
    """
    Moves zeroes to the end of the array in-place.
    """
    left = 0
    for right in range(len(nums)):
        if nums[right] != 0:
            nums[left], nums[right] = nums[right], nums[left]
            left += 1
    
    print(f"Result: {nums}")
    return nums

# --- Test ---
arr = [0, 1, 0, 3, 12]
move_zeroes(arr)
````

## 2️⃣ Pattern: Sliding Window
**Problem:** Given an array `nums` consisting of `n` integers, and an integer `k`. Find a contiguous subarray whose length is equal to `k` that has the maximum average value.

**Approach:** Calculate the sum of the first window. Then slide it right by subtracting the trailing element and adding the leading element.

````python
def find_max_average(nums, k):
    """
    Finds max average of any contiguous subarray of size k.
    """
    current_sum = sum(nums[:k])
    max_sum = current_sum
    
    for i in range(k, len(nums)):
        current_sum = current_sum - nums[i - k] + nums[i]
        max_sum = max(max_sum, current_sum)
        
    result = max_sum / k
    print(f"Max Average: {result}")
    return result

# --- Test ---
nums = [1, 12, -5, -6, 50, 3]
find_max_average(nums, k=4)
````

## 3️⃣ Pattern: Prefix Sum + Hash Map
**Problem:** Given an array of integers `nums` and an integer `k`, return the total number of continuous subarrays whose sum equals to `k`.

**Approach:** Store running sums in a dictionary. If `running_sum - k` exists in the map, a valid subarray ends here.

````python
def subarray_sum(nums, k):
    """
    Counts continuous subarrays that sum to k.
    """
    count = 0
    current_sum = 0
    prefix_sums = {0: 1}
    
    for num in nums:
        current_sum += num
        diff = current_sum - k
        
        if diff in prefix_sums:
            count += prefix_sums[diff]
            
        prefix_sums[current_sum] = prefix_sums.get(current_sum, 0) + 1
        
    print(f"Total subarrays equaling {k}: {count}")
    return count

# --- Test ---
nums = [1, 1, 1, -1, 1, 2]
subarray_sum(nums, k=2)
````

def longest_harmonious_subarray(nums):

    max_len = 0
    
    for num in counts:
