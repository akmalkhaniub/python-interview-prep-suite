# Mastering Stdin/Stdout Parsing (Micro1 / HackerRank)

In some automated platforms like Micro1, HackerRank, or older coding platforms, you don't get a neat `def solve(nums):` signature. Instead, you must read from Standard Input (`sys.stdin`) and print to Standard Output (`sys.stdout`).

This notebook demonstrates how to simulate and handle that raw input parsing robustly.


## 1. Simulating Standard Input in a Notebook
Normally, you'd use `import sys\nlines = sys.stdin.read().splitlines()`. To practice in Jupyter, we will use Python's `io.StringIO` to mock the input stream.


```
import sys
import io

# Mocking raw input text exactly as the platform would pass it:
mock_input = """3
5
1 2 3 4 5
3
10 20 30
1
99
"""

# Replace sys.stdin with our mock
sys.stdin = io.StringIO(mock_input)
```

## 2. Parsing the Input
Pattern: Read the number of test cases `T`, then loop `T` times. In each loop, read the array size `N`, then read the `N` integers on the next line.


```
def solve_test_cases():
    # read sys.stdin again
    input_data = sys.stdin.read().split()
    if not input_data: return
    
    # input_data is a flat list of all tokens (words/numbers) in the input, ignoring whitespace
    # This is often the safest and easiest way to parse ints!
    
    # Iterator makes it easy to grab the 'next' token
    iterator = iter(input_data)
    
    try:
        T = int(next(iterator))
        for _ in range(T):
            N = int(next(iterator))
            arr = [int(next(iterator)) for _ in range(N)]
            
            # Example operation: Print the sum of the array
            total = sum(arr)
            print(total)
    except StopIteration:
        pass

# Run the solver
solve_test_cases()
```

## 3. Alternative: Line-by-Line Parsing
Sometimes input has spaces that matter (e.g., strings with spaces). `sys.stdin.readline()` is better here.


```
# Mocking new input
mock_lines = """2
hello world
micro1 ai interview
"""
sys.stdin = io.StringIO(mock_lines)

def solve_lines():
    # readline() includes the newline character at the end, so we use .strip()
    T_str = sys.stdin.readline().strip()
    if not T_str: return
    T = int(T_str)
    
    for _ in range(T):
        line = sys.stdin.readline().strip()
        # Example operation: Reverse the string words
        reversed_words = ' '.join(line.split()[::-1])
        print(reversed_words)

solve_lines()
```

## Key Takeaways
1. **Use `sys.stdin.read().split()`** when you just need a flat list of integers/words and line breaks don't matter. It handles multiple spaces and newlines flawlessly.
2. **Use `sys.stdin.readline().strip()`** when you are reading entire string sentences or when the structure strictly relies on line-breaks.
3. **Always `print()` your final answer** exactly as requested by the platform format (e.g. one answer per line).
