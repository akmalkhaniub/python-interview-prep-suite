# Task Scheduler

**Problem Statement:**
Given a characters array `tasks`, representing the tasks a CPU needs to do, where each letter represents a different task. Tasks could be done in any order. Each task is done in one unit of time. For each unit of time, the CPU could complete either one task or just be idle.

However, there is a non-negative integer `n` that represents the cooldown period between two same tasks (the same letter in the array), that is that there must be at least `n` units of time between any two same tasks.

Return the least number of units of times that the CPU will take to finish all the given tasks.


## Approach: Greedy / Math (Optimal)
**Time Complexity:** O(N) where N is the number of tasks
**Space Complexity:** O(1) as there are at most 26 uppercase English letters

The most frequent task will require the most idle time. We can calculate the maximum possible idle slots and fill them with other tasks.


```
from collections import Counter

def leastInterval(tasks, n):
    task_counts = list(Counter(tasks).values())
    max_count = max(task_counts)
    
    # Count how many tasks have this maximum frequency
    max_count_tasks = task_counts.count(max_count)
    
    # We partition the schedule into blocks of size (n + 1)
    # The number of such blocks is (max_count - 1)
    part_count = max_count - 1
    part_length = n - (max_count_tasks - 1)
    empty_slots = part_count * part_length
    
    available_tasks = len(tasks) - max_count * max_count_tasks
    idles = max(0, empty_slots - available_tasks)
    
    return len(tasks) + idles

# Test
print('Least intervals:', leastInterval(['A','A','A','B','B','B'], 2)) # Expected 8 (A -> B -> idle -> A -> B -> idle -> A -> B)
```
