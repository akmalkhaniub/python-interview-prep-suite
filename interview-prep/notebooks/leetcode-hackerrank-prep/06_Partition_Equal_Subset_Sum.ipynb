# Partition Equal Subset Sum

**Problem Statement:**
Given an integer array `nums`, return `true` if you can partition the array into two subsets such that the sum of the elements in both subsets is equal or `false` otherwise.

This is a classic variation of the 0/1 Knapsack problem, very common when asked to divide items/tasks equally among two teams.


## Approach 1: Recursion with Memoization (Top-Down DP)
**Time Complexity:** O(N * Sum)
**Space Complexity:** O(N * Sum)
We can calculate the total sum. If it's odd, we can't partition it. If even, we look for a subset that sums up to `total_sum // 2`.


```
def canPartition_memo(nums):
    total_sum = sum(nums)
    if total_sum % 2 != 0: return False
    target = total_sum // 2
    memo = {}
    
    def dfs(i, current_sum):
        if current_sum == target: return True
        if i >= len(nums) or current_sum > target: return False
        if (i, current_sum) in memo: return memo[(i, current_sum)]
        
        memo[(i, current_sum)] = dfs(i + 1, current_sum + nums[i]) or dfs(i + 1, current_sum)
        return memo[(i, current_sum)]
        
    return dfs(0, 0)

# Test
print('Can partition:', canPartition_memo([1,5,11,5]))
```

## Approach 2: DP with 1D Array (Bottom-Up Optimal)
**Time Complexity:** O(N * Sum)
**Space Complexity:** O(Sum)
Instead of using recursion or a 2D array, we can use a 1D DP array or a set to keep track of all possible sums achievable so far.


```
def canPartition_dp(nums):
    total_sum = sum(nums)
    if total_sum % 2 != 0: return False
    target = total_sum // 2
    
    dp = set()
    dp.add(0)
    
    for num in nums:
        next_dp = set()
        for t in dp:
            if t + num == target: return True
            next_dp.add(t + num)
            next_dp.add(t)
        dp = next_dp
        
    return False

# Test
print('Can partition (Optimal DP):', canPartition_dp([1,5,11,5]))
```
