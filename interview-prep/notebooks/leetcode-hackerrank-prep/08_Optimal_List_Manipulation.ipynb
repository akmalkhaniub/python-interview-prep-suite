# Optimal List Manipulation Patterns

When platforms like Micro1 ask you to "process a list optimally", they are almost always looking for an $O(N)$ time complexity and $O(1)$ space complexity solution. 

This notebook covers the **Big 3** list manipulation patterns that turn $O(N^2)$ brute force approaches into $O(N)$ optimal solutions: **Two Pointers**, **Sliding Window**, and **Prefix Sums**.


## Pattern 1: Two Pointers (In-Place Modification)
**Scenario:** You need to filter, reverse, or shift elements without using extra memory.

**Example: Move Zeroes**
Given an integer array `nums`, move all `0`s to the end of it while maintaining the relative order of the non-zero elements.
*Constraint:* You must do this in-place without making a copy of the array.


```
def move_zeroes(nums):
    # left pointer tracks the position to place the next non-zero element
    left = 0
    
    # right pointer iterates through the array
    for right in range(len(nums)):
        if nums[right] != 0:
            # Swap elements
            nums[left], nums[right] = nums[right], nums[left]
            left += 1
            
    return nums

# Test
arr = [0, 1, 0, 3, 12]
print("Move Zeroes Result:", move_zeroes(arr))
```

## Pattern 2: Sliding Window
**Scenario:** You need to find a subarray (contiguous elements) that satisfies a condition (max sum, longest, shortest).

**Example: Maximum Average Subarray I**
Given an array `nums` consisting of `n` integers, and an integer `k`. Find a contiguous subarray whose length is equal to `k` that has the maximum average value.
*Bad approach:* Recalculating the sum of every window of size k takes $O(N \times K)$.
*Optimal approach:* Calculate the first window, then slide it by subtracting the outgoing element and adding the incoming element in $O(1)$ time.


```
def find_max_average(nums, k):
    # Calculate the sum of the very first window
    current_sum = sum(nums[:k])
    max_sum = current_sum
    
    # Slide the window across the rest of the array
    for i in range(k, len(nums)):
        current_sum = current_sum - nums[i - k] + nums[i]
        max_sum = max(max_sum, current_sum)
        
    return max_sum / k

# Test
print("Max Average:", find_max_average([1,12,-5,-6,50,3], 4)) # Expected: 12.75
```

## Pattern 3: Prefix Sum + Hash Map
**Scenario:** You need to find the number of subarrays (or a specific subarray) that sum to a target value, but the array might contain negative numbers (so sliding window won't work perfectly).

**Example: Subarray Sum Equals K**
Given an array of integers `nums` and an integer `k`, return the total number of continuous subarrays whose sum equals to `k`.
*Optimal approach:* Keep a running sum. Store how many times we've seen each running sum in a dictionary. If `running_sum - target` exists in our dictionary, we know a valid subarray ends here.


```
def subarray_sum(nums, k):
    count = 0
    current_sum = 0
    # Dictionary to store frequency of prefix sums.
    # Initialize with {0: 1} to handle the case where a prefix itself equals k.
    prefix_sums = {0: 1}
    
    for num in nums:
        current_sum += current_sum + num
        diff = current_sum - k
        
        # If diff is in our map, it means there is a subarray ending here that sums to k
        if diff in prefix_sums:
            count += prefix_sums[diff]
            
        # Add current sum to our map
        prefix_sums[current_sum] = prefix_sums.get(current_sum, 0) + 1
        
    return count

# Test
print("Subarrays equaling 2:", subarray_sum([1,1,1,-1,1,2], 2)) # Expected: 4
```
