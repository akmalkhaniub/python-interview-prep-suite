# Two Sum

**Problem Statement:**
Given an array of integers `nums` and an integer `target`, return indices of the two numbers such that they add up to `target`.

You may assume that each input would have exactly one solution, and you may not use the same element twice.


## Approach 1: Brute Force
**Time Complexity:** O(N^2)
**Space Complexity:** O(1)
Check every pair of elements to see if they sum to target.


```
def two_sum_brute(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]

# Test
print(two_sum_brute([2, 7, 11, 15], 9))
```

## Approach 2: Hash Map (Optimal)
**Time Complexity:** O(N)
**Space Complexity:** O(N)
Use a hash map to store the value and its index. For each element, check if `target - element` exists in the map.


```
def two_sum_optimal(nums, target):
    num_map = {}
    for i, num in enumerate(nums):
        diff = target - num
        if diff in num_map:
            return [num_map[diff], i]
        num_map[num] = i

# Test
print(two_sum_optimal([2, 7, 11, 15], 9))
```
