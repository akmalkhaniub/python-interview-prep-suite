# Valid Parentheses

**Problem Statement:**
Given a string `s` containing just the characters '(', ')', '{', '}', '[' and ']', determine if the input string is valid.
An input string is valid if:
1. Open brackets must be closed by the same type of brackets.
2. Open brackets must be closed in the correct order.


## Approach 1: String Replacement (Suboptimal)
**Time Complexity:** O(N^2) in worst case
**Space Complexity:** O(N)
Repeatedly replace '()', '{}', '[]' with empty string until string is empty or no more replacements can be made.


```
def is_valid_replace(s):
    while '()' in s or '{}' in s or '[]' in s:
        s = s.replace('()', '').replace('{}', '').replace('[]', '')
    return s == ''

# Test
print(is_valid_replace('({[]})'))
```

## Approach 2: Stack (Optimal)
**Time Complexity:** O(N)
**Space Complexity:** O(N)
Use a stack to keep track of opening brackets. When encountering a closing bracket, check if it matches the top of the stack.


```
def is_valid_stack(s):
    stack = []
    mapping = {')': '(', '}': '{', ']': '['}
    
    for char in s:
        if char in mapping:
            top_element = stack.pop() if stack else '#'
            if mapping[char] != top_element:
                return False
        else:
            stack.append(char)
            
    return not stack

# Test
print(is_valid_stack('({[]})'))
```
